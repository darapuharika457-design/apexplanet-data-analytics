## Step 8: Outlier Detection Using IQR Method

The Interquartile Range (IQR) method is used to identify extreme values (outliers) in the Sales colum

# Import visualization libraries


```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Create boxplot for Sales

plt.figure(figsize=(8,5))
sns.boxplot(x=df['Sales'])
plt.title("Boxplot of Sales")
plt.show()
```

### Sales Outlier Visualization

A boxplot was created to identify potential outliers in the Sales column.

Observations:
- Most sales values are concentrated in the lower range.
- Several extreme sales values are present.
- These extreme values appear as individual points beyond the whiskers of the boxplot.

Conclusion:
The Sales column contains potential outliers and will be further analyzed using the IQR method.

# Calculate first and third quartiles

```python
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)

# Calculate IQR

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
# Calculate lower and upper bounds

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
# Detect outliers

outliers = df[(df['Sales'] < lower_bound) | (df['Sales'] > upper_bound)]

print("Number of Outliers:", len(outliers))
```

```python
# Display first 5 outlier records

outliers.head()
```

### Outlier Analysis Result

The IQR method was applied to the Sales column.

Observations:
- Q1 = 17.28
- Q3 = 209.94
- IQR = 192.66
- Total Outliers = 1167

The lower and upper bounds were calculated and values outside these limits were identified as outliers.

This analysis helps in identifying unusually high or low sales transactions that may affect statistical analysis.

## Step 9: Outlier Removal

The identified outliers are removed from the Sales column using the IQR boundaries.

```python
# Remove outliers

df_clean = df[(df['Sales'] >= lower_bound) & (df['Sales'] <= upper_bound)]

print("Original Shape:", df.shape)
print("Cleaned Shape:", df_clean.shape)
```

```python
# Create boxplot after outlier removal

plt.figure(figsize=(8,5))
sns.boxplot(x=df_clean['Sales'])
plt.title("Boxplot of Sales After Outlier Removal")
plt.show()
```

## Step 10: Save Cleaned Dataset

The cleaned dataset is saved for future analysis.

```python
df_clean.to_csv("../data/SampleSuperstore_Cleaned.csv", index=False)

print("Cleaned dataset saved successfully.")
```
