```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Create boxplot for Sales

plt.figure(figsize=(8,5))
sns.boxplot(x=df['Sales'])
plt.title("Boxplot of Sales")
plt.show()
```

```python
Q1 = df['Sales'].quantile(0.25)   # Calculate the first quartile (Q1) and third quartile (Q3)
Q3 = df['Sales'].quantile(0.75)   # of the Sales column for outlier detection using the IQR method

# Calculate IQR 

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
# Calculate lower and upper bounds

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
# Detect outliers

outliers = df[(df['Sales'] < lower_bound) | (df['Sales'] > upper_bound)]

print("Number of Outliers:", len(outliers)) 
```

```python
# Display first 5 outlier records

outliers.head()
```

```python
# Remove outliers

df_clean = df[(df['Sales'] >= lower_bound) & (df['Sales'] <= upper_bound)]

print("Original Shape:", df.shape)
print("Cleaned Shape:", df_clean.shape)
```

```python
# Create boxplot after outlier removal

plt.figure(figsize=(8,5))
sns.boxplot(x=df_clean['Sales'])
plt.title("Boxplot of Sales After Outlier Removal")
plt.show()
```

```python
df_clean.to_csv("../data/SampleSuperstore_Cleaned.csv", index=False)    # Export the cleaned dataset to a CSV file and save it
                                                                        # in the data folder for future analysis and reporting

print("Cleaned dataset saved successfully.")
```
