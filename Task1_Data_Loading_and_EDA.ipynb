# Step 13: Outlier Detection Using IQR Method

This section identifies outliers in the Sales column using the Interquartile Range (IQR) technique.

```python
Q1 = df['Sales'].quantile(0.25)   # Calculate the first quartile (Q1) and third quartile (Q3)
Q3 = df['Sales'].quantile(0.75)   # of the Sales column for outlier detection using the IQR method

# Calculate IQR 

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
# Calculate lower and upper bounds

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
# Detect outliers

outliers = df[(df['Sales'] < lower_bound) | (df['Sales'] > upper_bound)]

print("Number of Outliers:", len(outliers)) 
```

```python
# Display first 5 outlier records

outliers.head()
```

```python
# Remove outliers

df_clean = df[(df['Sales'] >= lower_bound) & (df['Sales'] <= upper_bound)]

print("Original Shape:", df.shape)
print("Cleaned Shape:", df_clean.shape)
```

# Step 15: Exploratory Data Analysis (EDA) and Visualizations

This section creates visualizations to understand data distributions, category performance, correlations, and sales trends.

## Visualization 1: Sales Distribution

A histogram is used to understand the distribution of sales values and identify whether the data is normally distributed or skewed.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df['Sales'], bins=30)

plt.title('Distribution of Sales')
plt.xlabel('Sales')
plt.ylabel('Frequency')

plt.show()
```

### Observation

The Sales distribution is positively skewed. Most transactions have lower sales values, while only a small number of transactions contribute very high sales amounts. This indicates the presence of high-value sales outliers in the dataset.

## Visualization 2: Sales Outlier Analysis

A boxplot is used to detect outliers and understand the spread of sales values.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

sns.boxplot(x=df['Sales'])

plt.title('Boxplot of Sales')

plt.show()
```

### Observation

The boxplot reveals a large number of outliers in the Sales column. Most transactions have relatively low sales values, while a small number of transactions exhibit exceptionally high sales amounts. This confirms the presence of significant outliers identified using the IQR method.

## Visualization 3: Category-wise Sales Analysis

```python
plt.figure(figsize=(8,5))

df.groupby('Category')['Sales'].sum().plot(kind='bar')

plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')

plt.show()
```

### Observation

Technology generated the highest total sales among all categories, followed by Furniture and Office Supplies. This indicates that Technology products contribute the most to overall company revenue and represent a key business segment.

## Visualization 4: Category-wise Profit Analysis

This chart compares total profit generated by each product category.

```python
plt.figure(figsize=(8,5))

df.groupby('Category')['Profit'].sum().plot(kind='bar')

plt.title('Total Profit by Category')
plt.xlabel('Category')
plt.ylabel('Total Profit')

plt.show()
```

### Observation

Technology generated the highest profit among all categories, followed by Office Supplies. Although Furniture recorded strong sales, its profit was significantly lower, suggesting lower profit margins or higher operating costs within that category.

## Visualization 5: Correlation Heatmap

A heatmap is used to examine relationships between numerical variables in the dataset.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

sns.heatmap(
    df[['Sales','Quantity','Discount','Profit']].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')

plt.show()
```

### Observation

The correlation heatmap shows a moderate positive relationship between Sales and Profit (0.48), indicating that higher sales generally lead to higher profits. A negative correlation (-0.22) exists between Discount and Profit, suggesting that increased discounts tend to reduce profitability. Other variables show relatively weak correlations.

## Visualization 6: Sales vs Profit Relationship

A scatter plot is used to visualize the relationship between Sales and Profit.

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Sales'], df['Profit'])

plt.title('Sales vs Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')

plt.show()
```

### Observation

The scatter plot shows a moderate positive relationship between Sales and Profit. While higher sales generally lead to higher profits, several transactions generated losses despite having substantial sales values. This suggests that factors such as discounts, costs, or product margins may impact profitability.

## Visualization 7: Monthly Sales Trend

A line chart is used to analyze sales trends over time.

```python
monthly_sales = df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum()

monthly_sales.plot(figsize=(10,5))

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.show()
```

### Observation

The Monthly Sales Trend chart shows that sales fluctuate across different months and years. Several peaks and dips are observed, indicating seasonal variations in customer demand. Sales generally increase toward the end of the dataset period, with the highest sales recorded in late 2017. This suggests periods of strong business performance and opportunities for forecasting future sales trends.
