## SQL Fundamentals

# SELECT  
Used to choose which columns you want to see.

```python
query = """
SELECT "Customer Name",
        Sales
FROM superstore;
"""
pd.read_sql(query, conn)
```

# Insight:
 The SELECT statement is used to retrieve specific columns from the dataset.
 It helps focus on relevant information instead of viewing the entire table.

# WHERE
Filters data based on a condition

```python
query = """
SELECT *
FROM superstore
WHERE Category = 'Technology';
"""

pd.read_sql(query, conn)
```

# Insight:
 The WHERE clause filters records based on specific conditions.
 It helps analyze only the required subset of data, such as a product category or region.

# ORDER BY 
Sorts Data 

```python
query = """
SELECT "Customer Name",
       Sales
FROM superstore
ORDER BY Sales DESC;
"""

pd.read_sql(query, conn)
```

# Insight:
 ORDER BY arranges data in ascending or descending order.
 It helps identify top-performing customers, products, or regions.

# LIMIT
Restricts the number of rows returned

```python
query = """
SELECT *
FROM superstore
LIMIT 5;
"""
pd.read_sql(query, conn)
```

# Insight:
 LIMIT restricts the number of records returned by a query.
 It is useful for quickly inspecting sample data without loading the entire dataset.

# GROUP BY
Helps summarize the data

```python
query = """ 
SELECT Category,
       SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Category 
ORDER BY Total_Sales DESC;
"""
pd.read_sql(query,conn)
```

# Insight:
 GROUP BY combines records with similar values into groups.
 It helps summarize data and calculate metrics such as total sales,
 average profit, and customer counts.

 ## HAVING Clause

```python
query = """
SELECT Category,
       SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Category
HAVING SUM(Sales) > 700000;
"""

pd.read_sql(query, conn)
```

# Insight:
 HAVING filters grouped results after aggregation.
 It helps identify groups that meet specific conditions,
 such as categories with high sales or profit.

 # INNER JOIN

```python
customer_data = {
    "Customer_ID": ["C1", "C2", "C3"],
    "Customer_Name": ["Robert", "Nick", "John"]
}

customer_df = pd.DataFrame(customer_data)
customer_df.to_sql("customers", conn, if_exists="replace", index=False)
```

```python
query = """
SELECT customers.Customer_Name,
       orders.Sales
FROM customers
INNER JOIN orders
ON customers.Customer_ID = orders.Customer_ID;
"""

pd.read_sql(query, conn)
```

# Insight:
 INNER JOIN combines matching records from two tables.
It helps integrate related information stored across multiple tables.

# LEFT JOIN

```python
# left join returns all records from the left table and matching from the right table
```

```python
query = """
SELECT customers.Customer_Name,
       orders.Sales
FROM customers
LEFT JOIN orders
ON customers.Customer_ID = orders.Customer_ID;
"""

pd.read_sql(query, conn)
```

# Insight:
 LEFT JOIN returns all records from the left table and matching records
 from the right table. It helps identify missing or unmatched data.

 ## RIGHT JOIN

```python
# Right join returns all records from the right table and matching records from the left table
```

```python
query = """
SELECT customers.Customer_Name,
       orders.Sales
FROM orders
LEFT JOIN customers
ON orders.Customer_ID = customers.Customer_ID;
"""

pd.read_sql(query, conn)
```

# Insight:
 RIGHT JOIN returns all records from the right table and matching records
 from the left table. It is useful when all records from the right table
 need to be preserved.

 ## FULL OUTER JOIN

```python
# Full Outer JOin returns all records from both tables
```

```python
query = """
SELECT customers.Customer_Name,
       orders.Sales
FROM customers
LEFT JOIN orders
ON customers.Customer_ID = orders.Customer_ID

UNION

SELECT customers.Customer_Name,
       orders.Sales
FROM orders
LEFT JOIN customers
ON orders.Customer_ID = customers.Customer_ID;
"""

pd.read_sql(query, conn)
```

# Insight:
 FULL OUTER JOIN returns all records from both tables,
 whether matching or not. It provides a complete view of the data.

## Advanced SQL

# Subquery

```python
# A subquery is a query inside another query.
# It helps retrieve data based on the result of another query.
```

```python
query = """
SELECT "Customer Name",
       Sales
FROM superstore
WHERE Sales >
(
    SELECT AVG(Sales)
    FROM superstore
);
"""

pd.read_sql(query, conn)
```

# Insight:
 This query finds customers whose sales are greater than the average sales.
 Subqueries help compare records against aggregated values such as
 average sales, maximum profit, minimum sales, etc.

## CTE (WITH Clause)

```python
# CTE creates a temporary result set that can be reused in the main query
```

```python
query = """
WITH CategorySales AS
(
    SELECT Category,
           SUM(Sales) AS Total_Sales
    FROM superstore
    GROUP BY Category
)

SELECT *
FROM CategorySales;
"""

pd.read_sql(query, conn)
```

# Insight:
 Office Supplies generated the highest total sales,
 followed by Furniture and Technology.

## ROW_NUMBER() Function

```python
# Assign a unique row number based on sales ranking
```

```python
query = """
SELECT "Customer Name",
       Sales,

       ROW_NUMBER() OVER
       (
           ORDER BY Sales DESC
       ) AS Row_Number

FROM superstore
LIMIT 10;
"""

pd.read_sql(query, conn)
```

# Insight:
 ROW_NUMBER() assigns a unique sequential number to each row
 based on the specified sorting order. It is useful for ranking,
 identifying top records, and organizing data in a structured manner.

## RANK() Function

```python
# Rank customers based on sales. Same sales values get the same rank.
```

```python
query = """
SELECT "Customer Name",
       Sales,

       RANK() OVER
       (
           ORDER BY Sales DESC
       ) AS Sales_Rank

FROM superstore
LIMIT 10;
"""

pd.read_sql(query, conn)
```

# Insight:
 RANK() assigns the same rank to records with equal sales values.
 Useful for identifying top performers while handling ties correctly.

## LAG() Function

```python
# LAG() returns the previous row's sales value
```

```python
query = """
SELECT "Customer Name",
       Sales,

       LAG(Sales)
       OVER
       (
           ORDER BY Sales DESC
       ) AS Previous_Sales

FROM superstore
LIMIT 10;
"""

pd.read_sql(query, conn)
```

 # Insight:
 LAG() helps compare current sales with the previous sales value.
 Useful for trend analysis and performance comparison.

## LEAD() Function

```python
# LEAD() returns the next row's sales value
```

```python
query = """
SELECT "Customer Name",
       Sales,

       LEAD(Sales)
       OVER
       (
           ORDER BY Sales DESC
       ) AS Next_Sales

FROM superstore
LIMIT 10;
"""

pd.read_sql(query, conn)
```

# Insight:
 LEAD() helps compare current sales with the next sales value.
 Useful for identifying changes between consecutive records.

## VIEW (Reusable Query)

```python
# A VIEW is a virtual table created from a SQL query.
# It stores the query logic and can be reused like a normal table.
```

```python
# Create View
```

```python
query = """
DROP VIEW IF EXISTS High_Sales;
"""

conn.execute(query)
```

```python
query = """
CREATE VIEW High_Sales AS

SELECT "Customer Name",
       Category,
       Sales

FROM superstore

WHERE Sales > 400;
"""

conn.execute(query)
```

```python
# Retrieve all records from the High_Sales view.
# This view contains customers whose sales are greater than 400.
```

```python
query = """
SELECT *
FROM High_Sales;
"""

pd.read_sql(query, conn)

```

# Insight:
 The High_Sales view contains customers whose sales exceed 400.
 Views help simplify complex queries and improve reusability.
